# Hybrid Quantum–Spiking LeNet-5 on MNIST (rate-encoded)

A spiking LeNet-5 whose **F6 layer (120→84) is replaced by a variational quantum circuit (VQC)**, evaluated at every simulation time step inside the spiking pipeline.

```
 image ──rate code──► spikes (T steps)
   C1 conv5×5(6) → LIF → maxpool          6×14×14   ┐
   C3 conv5×5(16) → LIF → maxpool        16×5×5     │ classical spiking
   C5 FC 400→120 → LIF                    120       ┘
   Q  FC 120→n_q → angle map → VQC(n_q qubits, L layers) → ⟨Z_i⟩   n_q   ← quantum (replaces F6)
   F7 FC n_q→10 → LIF  → spike count over T → class
```

**Quantum layer** (per time step, per sample):
$$|\psi(\mathbf{x})\rangle=\prod_{l=1}^{L}\Big[\mathrm{CNOT}_{\text{ring}}\ \bigotimes_{i} R_Y(\theta_{l,i})R_Z(\phi_{l,i})\ \underbrace{\bigotimes_i R_Y(\pi\tanh x_i)}_{\text{data (re-)upload}}\Big]|0\rangle^{\otimes n_q},\qquad q_i=\langle\psi|Z_i|\psi\rangle\in[-1,1]$$

* LIF: $U[t]=\beta U[t-1]+I[t]$, $S[t]=\Theta(U[t]-\vartheta)$, soft reset $U\!\leftarrow\!U-S\vartheta$; fast-sigmoid surrogate gradient.
* Quantum part is simulated exactly with a **batched, differentiable PyTorch statevector simulator** (GPU-friendly; gradients by autograd, identical to the analytic/parameter-shift gradient). An optional cell cross-checks it against PennyLane.
* Optional **finite-shot** estimation of $\langle Z_i\rangle$ at evaluation time to emulate hardware sampling noise.
* A classical spiking LeNet-5 baseline (same pipeline, F6 classical) can be trained for comparison.

## 4. Rate encoder

````python
def rate_encode(x, T, gain=1.0):
    """(B,C,H,W) in [0,1] -> (T,B,C,H,W) Bernoulli spikes with p = gain*x."""
    p = (x * gain).clamp(0.0, 1.0)
    return torch.bernoulli(p.unsqueeze(0).expand(T, *p.shape))

img, lbl = test_ds[0]
spk = rate_encode(img.unsqueeze(0), CFG["T"], CFG["gain"])[:, 0, 0]
fig, ax = plt.subplots(1, 6, figsize=(14, 2.6))
ax[0].imshow(img[0], cmap="gray"); ax[0].set_title(f"input (label {lbl})")
for i, t in enumerate([0, 1, 2, 3]):
    ax[i+1].imshow(spk[t], cmap="gray"); ax[i+1].set_title(f"spikes t={t}")
ax[5].imshow(spk.mean(0), cmap="gray"); ax[5].set_title(f"mean rate, T={CFG['T']}")
for a in ax: a.axis("off")
plt.tight_layout(); plt.show()
````

## 5. Surrogate spike function and LIF neuron

````python
class FastSigmoidSpike(torch.autograd.Function):
    slope = CFG["slope"]
    @staticmethod
    def forward(ctx, x):
        ctx.save_for_backward(x)
        return (x > 0).float()
    @staticmethod
    def backward(ctx, g):
        (x,) = ctx.saved_tensors
        return g / (FastSigmoidSpike.slope * x.abs() + 1.0) ** 2

spike_fn = FastSigmoidSpike.apply


class LIF(nn.Module):
    def __init__(self, beta=0.9, threshold=1.0):
        super().__init__()
        self.beta, self.threshold, self.mem = beta, threshold, None
    def reset_state(self):
        self.mem = None
    def forward(self, x):
        if self.mem is None or self.mem.shape != x.shape:
            self.mem = torch.zeros_like(x)
        self.mem = self.beta * self.mem + x
        spk = spike_fn(self.mem - self.threshold)
        self.mem = self.mem - spk * self.threshold      # soft reset
        return spk
````

## 6. Quantum layer: batched differentiable statevector simulator

Convention: wire 0 is the most significant bit (same as PennyLane). The state is a complex tensor of shape `(B, 2**n)`.
* single-qubit gate on wire $w$: reshape to `(B, 2**w, 2, 2**(n-w-1))` and contract the middle axis, batched for data-dependent gates.
* CNOT ring $(0\to1)(1\to2)\dots(n{-}1\to0)$: a fixed basis permutation, precomputed once.
* $\langle Z_i\rangle = \sum_k |\psi_k|^2 (1-2\,b_i(k))$: one matrix product.

````python
def _ry(theta):
    """RY gate(s); theta (...,) -> (...,2,2) complex."""
    c, s = torch.cos(theta / 2), torch.sin(theta / 2)
    return torch.stack([torch.stack([c, -s], -1), torch.stack([s, c], -1)], -2).to(torch.complex64)

def _rz(phi):
    e = torch.exp(-0.5j * phi.to(torch.complex64))
    z = torch.zeros_like(e)
    return torch.stack([torch.stack([e, z], -1), torch.stack([z, e.conj()], -1)], -2)

def apply_1q(state, U, wire, n):
    """state (B,2**n); U (2,2) shared or (B,2,2) per-sample."""
    B = state.shape[0]
    s = state.view(B, 2 ** wire, 2, 2 ** (n - wire - 1))
    s = torch.einsum("ij,bajc->baic", U, s) if U.dim() == 2 else torch.einsum("bij,bajc->baic", U, s)
    return s.reshape(B, 2 ** n)


class VQCLayer(nn.Module):
    """Angle-encoded, hardware-efficient VQC with optional data re-uploading.
    input  x (B, n)  ->  output <Z_i> (B, n) in [-1, 1]."""

    def __init__(self, n_qubits=6, n_layers=3, reupload=True, shots=None):
        super().__init__()
        self.n, self.L, self.reupload, self.shots = n_qubits, n_layers, reupload, shots
        self.theta = nn.Parameter(0.1 * math.pi * torch.randn(n_layers, n_qubits))  # RY angles
        self.phi   = nn.Parameter(0.1 * math.pi * torch.randn(n_layers, n_qubits))  # RZ angles

        n, dim = n_qubits, 2 ** n_qubits
        bits = [[(k >> (n - 1 - i)) & 1 for i in range(n)] for k in range(dim)]
        # CNOT ring as basis permutation: new[f(k)] = old[k]  ->  new = old[:, f^{-1}]
        def ring(b):
            b = list(b)
            if n > 1:
                for c in range(n):
                    t = (c + 1) % n
                    if b[c]: b[t] ^= 1
            return b
        to_int = lambda b: int("".join(map(str, b)), 2)
        f = torch.tensor([to_int(ring(b)) for b in bits])
        self.register_buffer("perm", torch.argsort(f))
        self.register_buffer("zsign", 1.0 - 2.0 * torch.tensor(bits, dtype=torch.float32))  # (dim, n)

    def encode(self, state, angles):
        for w in range(self.n):
            state = apply_1q(state, _ry(angles[:, w]), w, self.n)
        return state

    def forward(self, x):
        B, n = x.shape[0], self.n
        angles = math.pi * torch.tanh(x)                      # map R -> (-pi, pi)
        state = torch.zeros(B, 2 ** n, dtype=torch.complex64, device=x.device)
        state[:, 0] = 1.0
        for l in range(self.L):
            if l == 0 or self.reupload:
                state = self.encode(state, angles)
            U = _ry(self.theta[l]) @ _rz(self.phi[l])           # (n,2,2): RZ then RY
            for w in range(n):
                state = apply_1q(state, U[w], w, n)
            state = state[:, self.perm]                        # CNOT ring
        probs = state.real ** 2 + state.imag ** 2
        z = probs @ self.zsign                                 # exact <Z_i>
        if self.shots and not self.training:                   # finite-shot emulation
            p1 = ((1 - z) / 2).clamp(0, 1)
            z = 1 - 2 * torch.distributions.Binomial(self.shots, p1).sample() / self.shots
        return z

    def extra_repr(self):
        return f"n_qubits={self.n}, layers={self.L}, reupload={self.reupload}, shots={self.shots}"
````

### 6a. Circuit schematic

````python
def draw_circuit(n, L, reupload):
    rows = [f"q{i}: |0>" for i in range(n)]
    for l in range(L):
        for i in range(n):
            enc = f"─RY(x{i})" if (l == 0 or reupload) else "─────────"
            rows[i] += f"{enc}─RZ(φ{l}{i})─RY(θ{l}{i})─"
        rows = [r + f"[CNOT ring L{l}]" for r in rows]
    for i in range(n): rows[i] += "─⟨Z⟩"
    print("\n".join(rows))
    print(f"\ntrainable circuit params: {2 * n * L}   Hilbert dim: {2 ** n}")
draw_circuit(CFG["n_qubits"], CFG["q_layers"], CFG["reupload"])
````

### 6b. Validation of the simulator against PennyLane (optional)

````python
try:
    import pennylane as qml
    vqc = VQCLayer(4, 2, reupload=True)
    n, L = 4, 2
    dev = qml.device("default.qubit", wires=n)
    @qml.qnode(dev, interface="torch")
    def circuit(x, theta, phi):
        a = math.pi * torch.tanh(x)
        for l in range(L):
            for i in range(n): qml.RY(a[i], wires=i)
            for i in range(n): qml.RZ(phi[l, i], wires=i); qml.RY(theta[l, i], wires=i)
            for c in range(n): qml.CNOT(wires=[c, (c + 1) % n])
        return [qml.expval(qml.PauliZ(i)) for i in range(n)]

    x = torch.randn(5, n)
    ours = vqc(x.float()).detach()
    ref = torch.stack([torch.stack(circuit(xi, vqc.theta.detach(), vqc.phi.detach())) for xi in x]).float()
    print("max |ours - pennylane| =", (ours - ref).abs().max().item())

    # gradient check: autograd (ours) vs PennyLane w.r.t. circuit parameters
    th = vqc.theta.detach().clone().requires_grad_(True)
    loss_ref = torch.stack(circuit(x[0], th, vqc.phi.detach())).sum(); loss_ref.backward()
    vqc.zero_grad(); vqc(x[:1].float()).sum().backward()
    print("max |grad diff| (theta) =", (vqc.theta.grad - th.grad.float()).abs().max().item())
except ImportError:
    print("PennyLane not installed - cross-check skipped.")
````

## 7. Hybrid Quantum–Spiking LeNet-5

````python
class HybridQSpikingLeNet5(nn.Module):
    LAYERS = ["C1", "C3", "C5", "F7-out"]

    def __init__(self, n_qubits=6, q_layers=3, reupload=True, shots=None,
                 num_classes=10, beta=0.9, threshold=1.0):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 6, 5, padding=2)
        self.conv2 = nn.Conv2d(6, 16, 5)
        self.fc1   = nn.Linear(16 * 5 * 5, 120)
        self.q_in  = nn.Linear(120, n_qubits)                       # classical -> qubit angles
        self.vqc   = VQCLayer(n_qubits, q_layers, reupload, shots)   # quantum F6
        self.fc_out = nn.Linear(n_qubits, num_classes)
        self.lifs = nn.ModuleList([LIF(beta, threshold) for _ in range(4)])

    def reset_state(self):
        for l in self.lifs: l.reset_state()

    def step(self, x):
        l1, l2, l3, lo = self.lifs
        s1 = F.max_pool2d(l1(self.conv1(x)), 2)
        s2 = F.max_pool2d(l2(self.conv2(s1)), 2)
        s3 = l3(self.fc1(s2.flatten(1)))
        q  = self.vqc(self.q_in(s3))                 # <Z_i> in [-1,1]
        so = lo(self.fc_out(q))
        return so, (s1, s2, s3, so), q

    def forward(self, spikes_in, record=False):
        self.reset_state()
        T = spikes_in.shape[0]
        counts, rates, out_train, q_trace = 0.0, [0.0] * 4, [], []
        for t in range(T):
            so, layers, q = self.step(spikes_in[t])
            counts = counts + so
            if record:
                out_train.append(so.detach()); q_trace.append(q.detach())
                for i, s in enumerate(layers):
                    rates[i] += s.detach().mean().item() / T
        if record:
            return counts, rates, torch.stack(out_train), torch.stack(q_trace)
        return counts


model = HybridQSpikingLeNet5(CFG["n_qubits"], CFG["q_layers"], CFG["reupload"], CFG["shots"],
                             beta=CFG["beta"], threshold=CFG["threshold"]).to(device)
n_q = sum(p.numel() for p in model.vqc.parameters())
n_c = sum(p.numel() for p in model.parameters()) - n_q
print(model)
print(f"Classical params: {n_c:,} | Quantum params: {n_q} | Total: {n_c + n_q:,}")
````

## 8. Classical spiking LeNet-5 baseline (for comparison)

````python
class SpikingLeNet5(nn.Module):
    LAYERS = ["C1", "C3", "C5", "F6", "F7-out"]
    def __init__(self, num_classes=10, beta=0.9, threshold=1.0):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 6, 5, padding=2); self.conv2 = nn.Conv2d(6, 16, 5)
        self.fc1 = nn.Linear(400, 120); self.fc2 = nn.Linear(120, 84); self.fc3 = nn.Linear(84, num_classes)
        self.lifs = nn.ModuleList([LIF(beta, threshold) for _ in range(5)])
    def reset_state(self):
        for l in self.lifs: l.reset_state()
    def forward(self, spikes_in, record=False):
        self.reset_state()
        l1, l2, l3, l4, lo = self.lifs
        counts, rates, out_train = 0.0, [0.0] * 5, []
        T = spikes_in.shape[0]
        for t in range(T):
            s1 = F.max_pool2d(l1(self.conv1(spikes_in[t])), 2)
            s2 = F.max_pool2d(l2(self.conv2(s1)), 2)
            s3 = l3(self.fc1(s2.flatten(1))); s4 = l4(self.fc2(s3)); so = lo(self.fc3(s4))
            counts = counts + so
            if record:
                out_train.append(so.detach())
                for i, s in enumerate((s1, s2, s3, s4, so)): rates[i] += s.detach().mean().item() / T
        return (counts, rates, torch.stack(out_train), None) if record else counts
````

### 8a. Width-matched classical control
Identical to the hybrid model except the VQC is swapped for a classical layer with the same interface:
$n_q$ inputs → $\tanh$ → FC($n_q$→$n_q$) → $\tanh$ → $n_q$ outputs in $[-1,1]$ ($n_q^2+n_q$ params vs. $2n_qL$ for the VQC).
This isolates the effect of the **quantum** layer from the effect of the **narrow bottleneck**.

````python
class ClassicalBottleneck(nn.Module):
    """Drop-in classical replacement for VQCLayer (same in/out shape and range)."""
    def __init__(self, n):
        super().__init__(); self.lin = nn.Linear(n, n); self.shots = None
    def forward(self, x):
        return torch.tanh(self.lin(torch.tanh(x)))

def make_classical_control():
    m = HybridQSpikingLeNet5(CFG["n_qubits"], CFG["q_layers"], CFG["reupload"], None,
                             beta=CFG["beta"], threshold=CFG["threshold"])
    m.vqc = ClassicalBottleneck(CFG["n_qubits"])
    return m.to(device)
````

## 9. Training and evaluation loops (shared by both models)

````python
def train_epoch(model, loader, opt):
    model.train()
    tot, correct, n = 0.0, 0, 0
    for x, y in tqdm(loader, leave=False, desc="train"):
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        counts = model(rate_encode(x, CFG["T"], CFG["gain"]))
        loss = F.cross_entropy(counts, y)            # spike counts as logits (rate decoding)
        opt.zero_grad(set_to_none=True)
        loss.backward()
        opt.step()
        tot += loss.item() * y.size(0); correct += (counts.argmax(1) == y).sum().item(); n += y.size(0)
    return tot / n, 100.0 * correct / n


@torch.no_grad()
def evaluate(model, loader, return_preds=False):
    model.eval()
    tot, correct, n, rate_sum, preds, targets = 0.0, 0, 0, None, [], []
    for x, y in tqdm(loader, leave=False, desc="eval"):
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        counts, rates, _, _ = model(rate_encode(x, CFG["T"], CFG["gain"]), record=True)
        tot += F.cross_entropy(counts, y).item() * y.size(0)
        p = counts.argmax(1); correct += (p == y).sum().item(); n += y.size(0)
        rate_sum = np.array(rates) if rate_sum is None else rate_sum + np.array(rates)
        if return_preds: preds.append(p.cpu()); targets.append(y.cpu())
    out = dict(loss=tot / n, acc=100.0 * correct / n, rates=rate_sum / len(loader))
    if return_preds:
        out["preds"], out["targets"] = torch.cat(preds).numpy(), torch.cat(targets).numpy()
    return out


def fit(model, opt, epochs, ckpt, tag):
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    hist, best = dict(train_loss=[], train_acc=[], test_loss=[], test_acc=[], rates=[], time=[]), 0.0
    for ep in range(1, epochs + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_epoch(model, train_dl, opt)
        te = evaluate(model, test_dl)
        sched.step()
        for k, v in zip(["train_loss", "train_acc", "test_loss", "test_acc", "rates", "time"],
                        [tr_loss, tr_acc, te["loss"], te["acc"], te["rates"], time.time() - t0]):
            hist[k].append(v)
        if te["acc"] > best:
            best = te["acc"]; torch.save(model.state_dict(), ckpt)
        print(f"[{tag}] Ep {ep:02d} | train {tr_loss:.4f} / {tr_acc:.2f}% | test {te['loss']:.4f} / {te['acc']:.2f}% "
              f"| best {best:.2f}% | rates [{', '.join(f'{r:.3f}' for r in te['rates'])}] | {hist['time'][-1]:.1f}s")
    print(f"[{tag}] best test accuracy: {best:.2f}%  -> {ckpt}")
    return hist, best
````

## 10. Train the hybrid model
Separate learning rates for classical weights and quantum circuit angles.

````python
q_params = list(model.vqc.parameters())
q_ids = {id(p) for p in q_params}
c_params = [p for p in model.parameters() if id(p) not in q_ids]
opt = torch.optim.Adam([{"params": c_params, "lr": CFG["lr"]},
                        {"params": q_params, "lr": CFG["q_lr"]}])
hist, best = fit(model, opt, CFG["epochs"], CFG["ckpt"], "Hybrid-Q")
````

## 13. Inside the quantum layer: ⟨Z⟩ traces and output spike rasters
